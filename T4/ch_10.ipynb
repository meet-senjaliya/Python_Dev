````python
import matplotlib.pyplot as plt
````

## Line Plot
```
plt.plot
Parameters-
x,y = data values
color = line color
linestyle = style of line
linewidth = thickness of line
marker = points symbol
markersize = size of marker
```

````python
plt.plot??
````

````python
x = [1,2.5,3,4]
y = [10,20,3,40]
plt.plot(
x,y,
color='m',
linestyle="--",
marker="^",
markersize=8

)
plt.title("A Line plot")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.grid(
True,
color='gray',
linestyle=":",
linewidth=1
)
plt.show()
````

````python
import numpy as np
````

````python
X=np.linspace(0,4*np.pi,1000)
Y=np.sin(X)
````

````python
fig,ax=plt.subplots()
ax.plot(X,Y)
fig.show
````

````python
X=np.linspace(0,10,100)
Y1,Y2=np.sin(X),np.cos(X)
ax.plot(X,Y1,X,Y2)

fig,(ax1,ax2)=plt.subplots(2,1)
ax1.plot(X,Y1,color="C1")
ax2.plot(X,Y2,color="C0")

fig,(ax1,ax2) = plt.subplots(1,2)
ax1.plot(Y1,X,color="C1")
ax2.plot(Y2,X,color="C0")
````

````python
x = [1,2,3,4]
y = [10,20,30,40]
print(x,y)
````

````python
plt.subplot(1,2,1)
plt.plot(x,y,color='b')
plt.title("X vs Y")

plt.subplot(1,2,2)
plt.plot(y,x,color='orange')
plt.title("Y vs X")
plt.show()
````

## Scatter plot

````python
plt.scatter(x,y,
color="b",
s=100,
alpha=1)#brightness

plt.grid()
plt.show()
````

## barplot

````python
plt.bar(
x,y,
color='g',
width=0.5
)
plt.show()
````

## barplot horizontal

````python
plt.barh(
x,y,
color='g',
height=0.5
)
plt.show()
````

## histogram

````python
data=[10,20,20,30,30,40]
plt.hist(
data,
bins=4,
orientation='vertical',
align='mid'

)
plt.show()
````

## pie chart

````python
labels=['A','B','C']
sizes = [25,35,15]
colors=['r','g','b']
plt.pie(
sizes,
labels=labels,
colors=colors,
startangle=90,
autopct="%1.2f%%",
explode=[1,0,0]
)
plt.show()
````
