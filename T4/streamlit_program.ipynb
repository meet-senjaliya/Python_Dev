```python
Create a Streamlit app where the user can input marks of 5 subjects (using number_input in columns).- Add a button to calculate:
  - Total marks
  - Average marks
  - Division (First/Second/Fail based on average)- Display results inside an expander section.
```

```python
%%writefile pb713.py
import streamlit as st
st.set_page_config(page_title="Hello Streamlit",layout="centered")
st.header("PB program 713")
c1,c2,c3,c4,c5 = st.columns(5)

m1=st.number_input("FCSP-1 ",0,100,0)
m2=st.number_input("FSD-1 ",0,100,0)
m3=st.number_input("DE ",0,100,0)
m4=st.number_input("PS",0,100,0)
m5=st.number_input("ETC",0,100,0)

if st.button("Calc Marks"):
    total = m1+m2+m3+m4+m5
    avg=total/5
    if avg>=60:
        division="First"
    elif avg>=40:
        division="Second"
    else:
        division="FAIL"
    with st.expander("Result"):
        st.write("Total marks",total)
        st.write("avg marks",avg)
        st.write("division ",division)
```

```python
!python -m streamlit run pb713.py
```

```python
%%writefile pb714.py
import streamlit as st
st.header("PB 714")
weight = st.number_input("weight(kg)",min_value=1.0)
height = st.number_input("height(cm)",min_value=1.0)

if st.button("calc BMI"):
    bmi = weight / ((height/100)**2)
    st.write(f"BMI VALUE  {bmi:.2f}")
    
    if bmi<18.5:
        st.warning("underwright")
    elif bmi<25.0:
        st.success("normal")
    elif bmi<30:
        st.warning("Overweight")
    else:
        st.error("obese")
```

```python
%%writefile pb715.py

import streamlit as st
import matplotlib.pyplot as plt
import numpy as np

st.header("PB 715")
n = st.number_input("enter no. of points to plot",min_value=10,step=10)

if st.button("genrate scatter plot"):
    x = np.random.rand(n)
    y = np.random.rand(n)
    
    plt.figure()
    plt.scatter(x,y)
    plt.xlabel("X values")
    plt.ylabel("Y values")
    plt.title("random scartter plot")
    st.pyplot(plt)
```
